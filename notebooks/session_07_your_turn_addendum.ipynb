# Session7 — additional Your turn checks

Explicit modifications from the supplied notebook: Exercise7 untuned baseline C=0.001; Exercise9 concatenate30 real public-data features and5000 synthetic noise features. AI-assisted execution; public teaching data only. The p-value is descriptive because resampled test sets overlap.

```python
from pathlib import Path
import json
import numpy as np
from scipy import stats
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split,StratifiedKFold,cross_val_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.feature_selection import SelectKBest,f_classif
X,y=load_breast_cancer(return_X_y=True)
fancy_scores=[];weak_scores=[]
for seed in range(10):
    Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=0.25,stratify=y,random_state=seed)
    fancy=make_pipeline(StandardScaler(),GradientBoostingClassifier(n_estimators=200,max_depth=3,random_state=seed)).fit(Xtr,ytr)
    weak=make_pipeline(StandardScaler(),LogisticRegression(C=0.001,max_iter=5000,random_state=seed)).fit(Xtr,ytr)
    fancy_scores.append(float(fancy.score(Xte,yte)));weak_scores.append(float(weak.score(Xte,yte)))
fancy_scores=np.array(fancy_scores);weak_scores=np.array(weak_scores)
delta=fancy_scores-weak_scores
naive_p=float(stats.ttest_rel(fancy_scores,weak_scores).pvalue)
print('Exercise7 Your turn: deliberately weak C=0.001 baseline')
print('GBM mean and sample SD:',fancy_scores.mean(),fancy_scores.std(ddof=1))
print('Weak baseline mean and sample SD:',weak_scores.mean(),weak_scores.std(ddof=1))
print('Mean paired gap and illustrative paired p:',delta.mean(),naive_p)
rng=np.random.default_rng(42)
noise=rng.normal(size=(len(y),5000));combined=np.column_stack([X,noise])
cv=StratifiedKFold(5,shuffle=True,random_state=42)
# Intentionally leaky feature selection; kept separate from the clean teaching branch.
selected=SelectKBest(f_classif,k=20).fit_transform(combined,y)
leaky=cross_val_score(LogisticRegression(max_iter=5000),selected,y,cv=cv,scoring='accuracy')
clean=cross_val_score(make_pipeline(SelectKBest(f_classif,k=20),LogisticRegression(max_iter=5000)),combined,y,cv=cv,scoring='accuracy')
print('Exercise9 Your turn:30 measured public-data features plus5000 noise features')
print('Leaky mean and sample SD:',leaky.mean(),leaky.std(ddof=1))
print('Clean mean and sample SD:',clean.mean(),clean.std(ddof=1))
result={'data_origin':'PUBLIC_WISCONSIN_DATA_PLUS_SYNTHETIC_NOISE','exercise7':{'seeds':list(range(10)),'fancy_scores':fancy_scores.tolist(),'weak_scores':weak_scores.tolist(),'fancy_mean':float(fancy_scores.mean()),'weak_mean':float(weak_scores.mean()),'paired_gap':float(delta.mean()),'naive_paired_p':naive_p},'exercise9':{'measured_features':30,'noise_features':5000,'leaky_scores':leaky.tolist(),'clean_scores':clean.tolist(),'leaky_mean':float(leaky.mean()),'clean_mean':float(clean.mean())},'interpretation':'A weak comparator can manufacture apparent superiority. Informative variables can mask leakage; a small gap does not validate the leaky branch. Overlapping resamples make the naive p-value illustrative, not independent replication. No mine data are involved.'}
Path('results.json').write_text(json.dumps(result,indent=2)+'\n',encoding='utf-8')
```
